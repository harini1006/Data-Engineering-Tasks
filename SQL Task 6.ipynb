#### 21.Display the details of the software developed in DBASE by Male Programmers, who belong to the institute in which most number of Programmers studied.  


```sql
SELECT S.*
FROM SOFTWARE S
JOIN PROGRAMMER P
ON S.PNAME = P.PNAME
JOIN STUDIES ST
ON P.PNAME = ST.PNAME
WHERE S.DEVELOPIN = 'DBASE'
  AND P.GENDER = 'M'
  AND ST.INSTITUTE = (
      SELECT INSTITUTE
      FROM STUDIES
      GROUP BY INSTITUTE
      ORDER BY COUNT(*) DESC
      LIMIT 1
  );
```

Create a T-SQL stored procedure named CalculateStoreRevenue that takes two parameters: @StartDate - Date (to specify the start date for the date range), @EndDate - Date (to specify the end date for the date range). The stored procedure should calculate and return the total revenue generated by each store within the specified date range. The result set should include the StoreName, Location, and TotalRevenue. Stored Procedure Requirements: The stored procedure should handle cases where there are no orders for a store within the specified date range (in such cases, the TotalRevenue should be 0). Use proper error handling to handle any potential errors or issues. Ensure that the stored procedure is optimized for performance.

**Create Store table and inser the values**

```sql
CREATE OR REPLACE TABLE Stores (
    StoreID INT PRIMARY KEY,
    StoreName VARCHAR(50),
    Location VARCHAR(50)
);
```

```sql
INSERT INTO Stores VALUES
(1,'D-Mart Poonamalle','Chennai'),
(2,'Commerical store Bg','Bangalore'),
(3,'KL 10','Hyderabad');
```

**Create Store orders table and insert the values**

```sql
CREATE OR REPLACE TABLE StoreOrders (
    OrderID INT PRIMARY KEY,
    StoreID INT,
    OrderDate DATE,
    Amount DECIMAL(10,2)
);
```

```sql
INSERT INTO StoreOrders VALUES
(101,1,'2026-04-01',5000),
(102,1,'2026-04-10',3000),
(103,2,'2026-04-05',7000);
```

**Create Stored procedure**

```sql
CREATE PROCEDURE CalculateStoreRevenue
    @StartDate DATE,
    @EndDate DATE
AS
BEGIN
    SET NOCOUNT ON;

    BEGIN TRY
        SELECT 
            S.StoreName,
            S.Location,
            ISNULL(SUM(O.Amount), 0) AS TotalRevenue
        FROM Stores S
        LEFT JOIN StoreOrders O
            ON S.StoreID = O.StoreID
            AND O.OrderDate BETWEEN @StartDate AND @EndDate
        GROUP BY S.StoreName, S.Location;
    END TRY
    BEGIN CATCH
        SELECT ERROR_MESSAGE() AS ErrorMessage;
    END CATCH
END;
```

#### 40.Merit is booking Rooms for Employees in a Hotel, who are interested in North India Holiday Vacation Trip. How can we create and utilize a User Defined Function in SQL Server to calculate the total cost of all hotel room reservations for a specific room type, taking into account the ticket type as a parameter input?. Retrieve the total cost of all reservations for ticket type ‘Standard. 


```sql
CREATE TABLE Reservations (
    ReservationID INT PRIMARY KEY,
    EmployeeName VARCHAR(50),
    RoomType VARCHAR(50),
    TicketType VARCHAR(50),
    Cost DECIMAL(10,2)
);
```

**Create a User defined Function to calculate total reservation cost**

```sql

CREATE FUNCTION TotalReservationCost (
    @RoomType VARCHAR(50),
    @TicketType VARCHAR(50)
)
RETURNS DECIMAL(10,2)

AS
BEGIN
    DECLARE @TotalCost DECIMAL(10,2);

    SELECT @TotalCost = SUM(Cost)
    FROM Reservations
    WHERE RoomType = @RoomType
      AND TicketType = @TicketType;

    RETURN ISNULL(@TotalCost, 0);
END;
```

```sql
SELECT task_schema.TotalReservationCost('Deluxe', 'Standard') AS TotalCost;
```
